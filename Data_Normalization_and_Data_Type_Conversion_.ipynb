# Part A - Data Normalization

```python
# Dataset
import pandas as pd
import numpy as no

data = {
    'Product' : ['Laptop','Mobile','iPad','Headphones','Smartwatch','Camera'],
    'Price' : [120000, 65000, 83000, 62500, 55000, 89000],
    'Units_sold' : [25,60,35,80,50,15],
    'Discount' : [10,15,12,20,18,8]
}
df = pd.DataFrame(data)
df
```

```python
# 1. Min-Max Normalization
# Min-Max normalization for the price

df['Price_MinMax'] = (df['Price'] - df['Price'].min()) / (df['Price'].max() - df['Price'].min())

print("\nMin-Max Normalizaed Price: ")
print(df[['Product','Price','Price_MinMax']])
```

```python
df['Dis_MinMax'] = (df['Discount'] - df['Discount'].min()) / (df['Discount'].max() - df['Discount'].min())

print("\nMin-Max Normalizaed Discount: ")
print(df[['Product','Discount','Dis_MinMax']])
```

```python
# Normalize multiple columns

cols = ['Price','Units_sold','Discount']
df_norm = (df[cols]-df[cols].min())/(df[cols].max()-df[cols].min())
df[cols] = df_norm
df
```

```python
# 2. Z-Score Normalization

df['Units_Zscore'] = (df['Units_sold'] - df['Units_sold'].mean()) / df['Units_sold'].std()

print("\nZ-Score Normalizaed Units: ")
print(df[['Product','Units_sold','Units_Zscore']])
```

```python
# 3. Decimal Scaling

df['Price_DS'] = df['Price'] / 100000

print("\nDecimal Scaling Normalizaed Price: ")
print(df[['Product','Price','Price_DS']])
```

# Part B - Turning categorical variables into quantitative variables in Pyhton

```python
data1 = {
    'Order_ID' : [101,102,103,104,105, 106],
    'Customer_gender' : ['Male','Female','Female','Male','Female','Male'],
    'Product_category' : ['Electronics','Clothing','Electronics','Clothing','Electronics','Clothing'],
    'Payment_method' : ['Credit Card','PayPal','PayPal','Credit Card','PayPal','Credit Card'],
    'City' : ['New York','Philadelphia','San Diego','Los Angeles','Boston', 'Chicago'],
    'Order_value' : [4500,2300,3200,5200,1800,4100]
}
df1 = pd.DataFrame(data1)
df1
```

```python
# Method 1: Label Encoding

from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df1['Gender_Label'] = le.fit_transform(df1['Customer_gender'])
print(df1)
```

```python
# Method 2: One-Hot Encoding

df_encoded = pd.get_dummies(df1, columns=['Payment_method'])
print(df_encoded)
```

```python
# Method 3 : Dummy Encoding (Drop first)

df_dummy = pd.get_dummies(df1, columns=['Product_category'], drop_first=True)
print(df_dummy)
```

```python
import pandas as pd
import numpy as np

df = pd.read_csv("/content/amazon_products_dataset_Expt-14.csv")
df
```

```python
# 1. Min-Max

df['Price_minmax'] = (df['Price'] - df['Price'].min()) / (df['Price'].max() - df['Price'].min())
print("\nMin-Max Normalized Price: ")
print(df[['Product_Name','Price','Price_minmax']])

df['Rating_minmax'] = (df['Rating'] - df['Rating'].min()) / (df['Rating'].max() - df['Rating'].min())
print("\nMin-Max Normalized Rating: ")
print(df[['Product_Name','Rating','Rating_minmax']])

df['Reviews_minmax'] = (df['Reviews'] - df['Reviews'].min()) / (df['Reviews'].max() - df['Reviews'].min())
print("\nMin-Max Normalized Reviews: ")
print(df[['Product_Name','Reviews','Reviews_minmax']])

df['Units_Sold_minmax'] = (df['Units_Sold'] - df['Units_Sold'].min()) / (df['Units_Sold'].max() - df['Units_Sold'].min())
print("\nMin-Max Normalized Units Sold: ")
print(df[['Product_Name','Units_Sold','Units_Sold_minmax']])
```

```python
# 2. Z Score Normalization

df['Price_zscore'] = (df['Price'] - df['Price'].mean()) / df['Price'].std()
print("\nZ-Score Normalized Price: ")
print(df[['Product_Name','Price','Price_zscore']])

df['Rating_zscore'] = (df['Rating'] - df['Rating'].mean()) / df['Rating'].std()
print("\nZ-Score Normalized Rating: ")
print(df[['Product_Name','Rating','Rating_zscore']])
```

```python
# 3. Decimal Scaling - Reviews

df['Reviews_ds'] = df['Reviews'] / 10000
print("\nDecimal Scaling Normalized Reviews: ")
print(df[['Product_Name','Reviews','Reviews_ds']])
```
