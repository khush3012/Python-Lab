```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
data = {
    'Days' : ['Mon','Tue','Wed','Thu','Fri'],
    'Study_Hours' : [2,3,6,1,3.5],
    'Marks' : [93,60,67,70,85],
    'Attendance': [60,70,65,80,75],
    'Assignments_Completed' : [1,2,1,3,2]

}
df = pd.DataFrame(data)
df
```

```python
# 1. Line Chart (Trend)

plt.plot(df['Days'],df['Study_Hours'], marker = 'o')
plt.title('Study Hours Over Days')
plt.xlabel('Days')
plt.ylabel('Study Hours')
plt.show()
```

```python
# Advanced Line Chart

plt.figure(figsize = (7,4))

plt.plot(df['Days'],df['Study_Hours'], marker='o',color='mediumspringgreen',label='Study Hours')

plt.plot(df['Days'],df['Marks'],marker='s',color='orange',label='Marks')

plt.title("Study Hours and Marks Trend")
plt.xlabel('Days')
plt.ylabel('Value')
plt.legend()
plt.show()
```

```python
# 2. Bar Chart (Comparision)

plt.bar(df['Days'],df['Marks'],color='hotpink')

plt.title("Marks Comparison")
plt.xlabel('Days')
plt.ylabel('Marks')
plt.show()
```

```python
plt.bar(df['Study_Hours'],df['Marks'], color = 'rebeccapurple')

plt.title("Study Hours vs Marks")
plt.xlabel('Study Hours')
plt.ylabel('Marks')
plt.show()
```

```python
# Advanced Bar Graph

bars = plt.bar(df['Days'],df['Marks'],color='skyblue')
for bar in bars:
  y = bar.get_height()
  plt.text(bar.get_x() + bar.get_width()/2,y,str(y),ha='center',va='bottom')

plt.title("Marks Comparison")
plt.xlabel('Days')
plt.ylabel('Marks')
plt.grid(axis='y')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd


df = pd.DataFrame(data)

x = np.arange(len(df['Days']))
width = 0.25
plt.bar(x-width/2, df['Study_Hours'], width,label = 'Study_Hours')
plt.bar(x+width/2, df['Marks'], width,label = 'Marks')
plt.xlabel('Days')
plt.ylabel('Value')
plt.title('Study Hours and Marks Comparison')
plt.legend()
plt.show()
```

```python
# 3. Histogram (Dsitribution)
# Shows how valuea are spread

data = { 'Marks': [50,60,55,70,65,75,80,85,90,45,67,72,77,82,88] }
df = pd.DataFrame(data)

plt.hist(df['Marks'], bins=3, edgecolor = 'magenta', alpha=0.7)
plt.title("Marks Distribution")
plt.xlabel('Marks')
plt.ylabel('Frequency')
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

data = {
    'Days' : ['Mon','Tue','Wed','Thu','Fri'],
    'Study_Hours' : [2,3,6,1,3.5],
    'Marks' : [93,60,67,70,85],
    'Attendance': [60,70,65,80,75],
    'Assignments_Completed' : [1,2,1,3,2]
}
df = pd.DataFrame(data)

# 4. Scatter Plot (Relationship)
plt.scatter(df['Study_Hours'], df['Marks'])

plt.title("Study Hours vs Marks")
plt.xlabel('Study Hours')
plt.ylabel('Marks')
plt.show()
```

```python
# Conditional scatter plot
# Example category

df['Result'] = ['Fail','Pass','Fail','Pass','Pass']
colors = ['lightcoral' if r=='Fail' else 'turquoise' for r in df['Result']]

plt.scatter(df['Study_Hours'],df['Marks'],c=colors)
plt.title("Study Hours vs Marks")
plt.xlabel('Study Hours')
plt.ylabel('Marks')
plt.show()
```

# Same graphs using Seaborn (Better Visualization)

```python
data = {
    "Day": ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun', 'Mon', 'Tue', 'Wed'],
    "Region": ['North', 'South', 'East', 'West', 'North', 'South', 'East', 'West', 'North', 'South'],
    "Sales": [200, 220, 250, 270, 300, 350, 400, 280, 260, 290],
    "Profit": [20, 25, 30, 35, 50, 60, 80, 40, 38, 45],
    "Customers": [30, 35, 40, 45, 50, 65, 80, 55, 48, 52],
    "Category": ['Food', 'Clothing', 'Electronics', 'Food', 'Clothing', 'Electronics', 'Food', 'Clothing', 'Electronics', 'Food']
}
df = pd.DataFrame(data)
df
```

```python
# Line Plot

sns.lineplot(x ='Day', y='Sales', data =df, color='khaki')
plt.title("Sales Over Days")
plt.xlabel('Day')
plt.ylabel('Sales')
plt.show()

```

```python
# Bar Chart

sns.barplot(x = 'Region', y = 'Profit', data = df)
plt.title("Profit by Region")
plt.xlabel('Region')
plt.ylabel('Profit')
plt.show()
plt
```

```python
# Scatter Plot

sns.scatterplot(x = 'Customers', y = 'Profit', data = df,color='greenyellow')
plt.title("Customers vs Profit")
plt.xlabel('Customers')
plt.ylabel('Profit')
plt.show()
plt
```

```python
# Histogram

sns.histplot(df['Sales'], bins = 3, kde = True,color='darksalmon')
plt.title("Sales Distribution")
plt.xlabel('Sales')
plt.ylabel('Frequency')
plt.show()
```
